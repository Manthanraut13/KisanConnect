# 02. Demand Forecasting using Meta Prophet
**Kisan Connect AI Service — SIH 2026**
Demonstrating time-series forecasting for 7-day crop price predictions.

```python
import pandas as pd
from prophet import Prophet

# Load crop data
df = pd.read_csv('../app/data/agmarknet_sample.csv', parse_dates=['date'])
crop_df = df[(df['commodity'] == 'Tomato') & (df['district'] == 'Nashik')]
prophet_df = crop_df[['date', 'modal_price']].rename(columns={'date': 'ds', 'modal_price': 'y'})
prophet_df['y'] = prophet_df['y'] / 100.0  # Rs/kg

# Fit Prophet
model = Prophet(yearly_seasonality=True, weekly_seasonality=True)
model.fit(prophet_df)

future = model.make_future_dataframe(periods=7)
forecast = model.predict(future)
print(forecast[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].tail(7))
```
