This notebook explores the 2.5-year historical price dataset across 20 crops and 10 districts.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv('../app/data/agmarknet_sample.csv', parse_dates=['date'])
print(f"Total records: {len(df):,}")
print(df.head())
```

```python
# Plot Tomato prices in Nashik over time
tomato_nashik = df[(df['commodity'] == 'Tomato') & (df['district'] == 'Nashik')].sort_values('date')
plt.figure(figsize=(12, 5))
plt.plot(tomato_nashik['date'], tomato_nashik['modal_price'] / 100.0, color='green', label='Price per Kg (₹)')
plt.title('Tomato Price Trend in Nashik (2024 - 2026)')
plt.xlabel('Date')
plt.ylabel('Price (₹/kg)')
plt.grid(True)
plt.legend()
plt.show()
```
